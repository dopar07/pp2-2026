# abs(), all(), any()
- 숫자는 `0`이 거짓, 나머지는 참이다.
- `all()`: 하나라도 거짓이면 `False`
- `any()`: 하나라도 참이면 `True`

````python
print(abs(-20), abs(-30.92))

print(all([1, 3, 4, 6]))            # 모든 값이 참이다.
print(all([1, 3, 4, 0]))            # 하나의 값이 거짓이다.
print(all([True, 0, False, 0]))     # 하나의 값만 참이다.

print(any([0, 1, 2, 3]))
print(any([0, False]))
````

# eval(), sum(), len(), list()

  - 입력받은 문자열을 그대로 실행하므로 믿을 수 없는 입력에는 쓰지 않는다.
- `sum(seq, 초기값)`: 두 번째 인수를 주면 그 값에서부터 더한다.

````python
x = 10
print(eval('x + 1'))

exp = "2**10"                       # exp = input("파이썬의 수식을 입력하시오: ")
print(eval(exp))

print(sum([1, 2, 3]))
print(sum([1, 2, 3], 20))           # 20 + 1 + 2 + 3

print(len("All's well that ends well. "))
print(len([1, 2, 3, 4, 5]))

print(list('abcdefg'))              # 문자열 -> 리스트
print(list((1, 2, 3, 4, 5, 6)))     # 튜플 -> 리스트
````

# map(), dir(), max(), min()
- `map(함수, 반복가능객체)`: 각 항목에 함수를 적용한다.
  - 결과는 리스트가 아니라 map 객체이므로 `list()`로 바꿔서 본다.
- `dir(객체)`: 객체가 가진 속성과 메소드의 이름을 보여 준다.

````python
def square(n):
    return n * n

mylist = [1, 2, 3, 4, 5]
result = map(square, mylist)
print(result)                       # map 객체
print(list(result))

print(dir([1, 2, 3])[:4], "...")    # 너무 길어서 앞의 4개만 출력
print([name for name in dir([1, 2, 3]) if not name.startswith("__")])

values = [1, 2, 3, 4, 5]
print(max(values), min(values))
print(max("python"), min("python")) # 문자열은 문자 코드 순서로 비교
````

# enumerate(), filter(), zip()
- `enumerate()`: 인덱스와 항목을 함께 꺼낼 때 사용한다. `start`로 시작 번호를 바꿀 수 있다.
- `filter(함수, 반복가능객체)`: 함수가 `True`를 반환하는 항목만 남긴다.
- `zip()`: 같은 위치의 항목끼리 튜플로 묶는다. **길이가 짧은 쪽**에 맞춰진다.

````python
seasons = ['Spring', 'Summer', 'Fall', 'Winter']
print(list(enumerate(seasons)))
print(list(enumerate(seasons, start=1)))
for index, season in enumerate(seasons):
    print(str(index), ":", season)

def myfilter(x):
    return x > 3

result = filter(myfilter, (1, 2, 3, 4, 5, 6))
print(list(result))

t1 = tuple(range(1, 7))
result2 = filter(lambda x: x > 4, t1)       # 람다식을 써도 된다.
print(list(result2))

numbers = [1, 2, 3, 4]
slist = ['one', 'two', 'three', 'four', 'five']
print(list(zip(numbers, slist)))            # 'five'는 짝이 없어서 빠진다.
print(dict(zip(numbers, slist)))            # 딕셔너리로 만들기

dict1 = {0: "영", 1: "일"}
print(list(zip(numbers, dict1)))            # 딕셔너리는 키가 묶인다.
````

# Lab: 내장 함수 예제
- 파티에 적합한 양의 음식을 구입하려면 총 인원이 몇 명인지 알아야 한다.
- `sum()`: 총 인원 / `any()`: 한 사람이라도 오는지 / `all()`: 모든 그룹이 오는지 / `max()`: 가장 많이 오는 그룹의 인원

````python
def party_info(invitations, persons):
    print("총 인원:", sum(persons))
    print("한 사람이라도 오는가?", any(persons))
    print("모든 그룹이 오는가?", all(persons))
    print("가장 많이 오는 그룹의 인원:", max(persons))
    for name, count in zip(invitations, persons):
        print(f"  {name}: {count}명")

def test_party_info():
    invitations = ["Kim", "Lee", "Park", "Choi"]
    persons = [1, 3, 0, 6]
    party_info(invitations, persons)

test_party_info()
````

# 정렬과 탐색
- `리스트.sort()`: 리스트의 메소드. **원래의 리스트**를 정렬된 상태로 변경한다. (반환값은 `None`)
- `sorted(시퀀스, key=함수, reverse=여부)`: 내장 함수. 반복 가능한 객체로부터 정렬된 **새로운 리스트**를 생성한다.
  - `key`: 각 원소의 크기(비교할 값)를 반환하는 함수
  - `reverse=True`: 내림차순

````python
print(sorted([4, 2, 3, 5, 1]))      # 새로운 리스트 생성

myList = [4, 2, 3, 5, 1]
print(myList.sort())                # 원래의 리스트를 정렬하고 None을 반환
print(myList)

# key 매개변수: 대소문자를 구분하지 않고 정렬
words = "The health know not of their health, but only the sick".split()
print(sorted(words))                # 대문자가 소문자보다 앞에 온다.
print(sorted(words, key=str.lower))
````

# key를 이용한 정렬
- 튜플이나 객체를 정렬할 때는 무엇을 기준으로 비교할지 `key`로 알려 준다.
- `key`에는 보통 람다식을 넣는다.

````python
students = [
    ('홍길동', 3.9, 20160303),
    ('김철수', 3.0, 20160302),
    ('최자영', 4.3, 20160301),
]
print(sorted(students, key=lambda student: student[2]))     # 학번 순

class Student:
    def __init__(self, name, grade, number):
        self.name = name
        self.grade = grade
        self.number = number

    def __repr__(self):
        return repr((self.name, self.grade, self.number))

students = [
    Student('홍길동', 3.9, 20160303),
    Student('김철수', 3.0, 20160302),
    Student('최자영', 4.3, 20160301),
]
print(sorted(students, key=lambda student: student.number))                 # 오름차순
print(sorted(students, key=lambda student: student.number, reverse=True))   # 내림차순
print(sorted(students, key=lambda student: student.grade, reverse=True))    # 학점이 높은 순
````

# Lab: 키를 이용한 정렬 예제
- 주소록을 작성한다. 사람을 `Person` 클래스로 나타낸다.
  - `name`: 이름 (문자열)
  - `age`: 나이 (정수형)
- 나이순으로 정렬하여 보여 준다. 정렬의 기준은 사람의 나이이다.

````python
class Person(object):
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return "<이름: %s, 나이: %s>" % (self.name, self.age)

    def getName(self):
        return self.name

# 정렬 함수
def keyAge(person):                 # 크기를 비교할 값 (나이)
    return person.age

def test_person_sort():
    people = [Person("홍길동", 20), Person("최자영", 38), Person("김철수", 35)]
    print(sorted(people, key=keyAge))                       # 나이순
    print(sorted(people, key=lambda x: x.getName()))        # 이름순

test_person_sort()
````

# 람다식
- 람다식(lambda expression)은 이름은 없고 몸체만 있는 함수이다.
- `lambda` 키워드로 만든다. → `lambda 매개변수: 수식`
- 딱 한 번 사용하는 함수를 만드는 데 사용된다.
- 수식 하나만 쓸 수 있고, 그 수식의 값이 반환값이 된다. (`return`을 쓰지 않는다.)

````python
f = lambda x, y: x + y
print("정수의 합 : ", f(10, 20))
print("정수의 합 : ", f(20, 20))

# 위의 람다식은 아래의 함수와 같다.
def get_sum(x, y):
    return x + y

print("정수의 합 : ", get_sum(10, 20))
print("정수의 합 : ", get_sum(20, 20))
````

# map(), filter(), reduce() 함수와 람다식
- `map(lambda x: 수식, 시퀀스)`: 각 항목을 수식으로 바꾼다.
- `filter(lambda x: 조건, 시퀀스)`: 조건이 참인 항목만 남긴다.
- `functools.reduce(func, seq)`: `func()` 함수를 시퀀스에 **연속적으로** 적용하여 단일 값을 반환한다.
  - `reduce(lambda x, y: x + y, [1, 2, 3, 4])` → `((1 + 2) + 3) + 4`

````python
import functools

list_a = [1, 2, 3, 4, 5]
f = lambda x: 2 * x
print(list(map(f, list_a)))

list_a = [1, 2, 3, 4, 5, 6]
print(list(filter(lambda x: x % 2 == 0, list_a)))

data = [(3, 100), (1, 200), (7, 300), (6, 400)]
sorted(data, key=lambda item: item[0])          # 반환값을 받지 않으면
print(data)                                     # data는 그대로이다.
print(sorted(data, key=lambda item: item[0]))

print(functools.reduce(lambda x, y: x + y, [1, 2, 3, 4]))
print(functools.reduce(lambda x, y: x * y, [1, 2, 3, 4]))   # 1*2*3*4
````

# Lab: 람다식으로 온도 변환하기
- 람다식과 `map()` 함수를 사용하여 화씨 온도를 섭씨 온도로 변환한다.
- 섭씨 = (5 / 9) × (화씨 − 32)

````python
# 람다식을 사용하지 않는 버전
def celsius(T):
    return (5.0 / 9.0) * (T - 32.0)

def test_celsius():
    f_temp = [0, 10, 20, 30, 40, 50]
    c_temp = map(celsius, f_temp)
    print(list(c_temp))

    # 람다식을 사용하는 버전
    c_temp = map(lambda x: (5.0 / 9.0) * (x - 32.0), f_temp)
    print(list(c_temp))

test_celsius()
````

- 아래의 데이터를 처리하여 (주문 번호, 총액) 튜플이 저장된 리스트를 반환한다.

| 주문 번호 | 물품 | 수량 | 품목 당 가격 |

| 2 | 셔츠 | 6 | 24000 |
| 3 | 바지 | 3 | 50000 |
| 4 | 코트 | 6 | 300000 |

````python
def order_totals(orders):
    return list(map(lambda x: (x[0], x[2] * x[3]), orders))

def test_order_totals():
    orders = [["1", "재킷", 5, 120000],
              ["2", "셔츠", 6, 24000],
              ["3", "바지", 3, 50000],
              ["4", "코트", 6, 300000]]
    result = order_totals(orders)
    print(result)
    print("전체 주문 금액:", functools.reduce(lambda x, y: x + y[1], result, 0))

test_order_totals()
````

# 이터레이터
- `for` 루프와 함께 사용할 수 있는 객체를 이터러블 객체(iterable object)라고 한다. (리스트, 튜플, 문자열, 딕셔너리, 파일 등)
- 객체가 이터레이터가 되려면 다음 두 메소드를 정의한다.
  - `__iter__()`: 이터레이터 객체 자신을 반환한다.
  - `__next__()`: 다음 반복을 위한 값을 반환한다. 더 이상 값이 없으면 `StopIteration` 예외를 발생시킨다.
- `for` 문은 내부적으로 `iter()`로 이터레이터를 얻고, `StopIteration`이 발생할 때까지 `next()`를 호출한다.

````python
it = iter([10, 20])                 # 리스트에서 이터레이터를 얻는다.
print(next(it))
print(next(it))
try:
    next(it)
except StopIteration:
    print("더 이상 값이 없다 -> StopIteration")

class MyCounter(object):
    # 생성자 메소드를 정의한다.
    def __init__(self, low, high):
        self.low = low
        self.current = low
        self.high = high

    # 이터레이터 객체로서 자신을 반환한다.
    def __iter__(self):
        return self

    def __len__(self):
        return self.high - self.low + 1

    def __next__(self):
        # current가 high보다 크면 StopIteration 예외를 발생한다.
        # current가 high보다 작거나 같으면 다음 값을 반환한다.
        if self.current > self.high:
            raise StopIteration
        else:
            self.current += 1
            return self.current - 1

c = MyCounter(1, 10)
print(len(c))
for i in c:
    print(i, end=' ')
print()
for i in c:                         # 이미 끝까지 간 이터레이터는 다시 돌지 않는다.
    print(i, end=' ')
````

# 제너레이터
- 제너레이터(generator)는 키워드 `yield`를 사용하여 **함수로부터 이터레이터를 생성**하는 방법이다.
- `yield`를 만나면 값을 내보내고 그 자리에서 멈췄다가, 다음 값을 요청받으면 멈춘 곳부터 이어서 실행한다.
- `__iter__()`, `__next__()`를 직접 만들지 않아도 되므로 이터레이터 클래스보다 간단하다.
- 값을 미리 다 만들어 두지 않고 필요할 때 하나씩 만들기 때문에 메모리를 아낄 수 있다.

````python
def myGenerator():
    yield 'first'
    yield 'second'
    yield 'third'

for word in myGenerator():
    print(word)

g = myGenerator()
print(type(g))
print(next(g), next(g))             # next()로 하나씩 꺼낼 수도 있다.

# MyCounter 클래스를 제너레이터로 다시 작성
def my_counter(low, high):
    current = low
    while current <= high:
        yield current
        current += 1

print(list(my_counter(1, 10)))
````

# Lab: 피보나치 이터레이터
- 피보나치 수열이란 앞의 두 수의 합이 바로 뒤의 수가 되는 수열이다. → `1 1 2 3 5 8 13 21 34`
- 피보나치 수열의 수들을 생성하는 이터레이터 클래스를 정의한다.
- 같은 일을 하는 제너레이터도 만들어서 비교해 본다.

````python
class FibIterator:
    def __init__(self, a=1, b=0, maxValue=50):
        self.a = a
        self.b = b
        self.maxValue = maxValue

    def __iter__(self):
        return self

    def __next__(self):
        n = self.a + self.b
        if n > self.maxValue:
            raise StopIteration()
        self.a = self.b
        self.b = n
        return n

# 제너레이터 버전
def fib_generator(maxValue=50):
    a, b = 1, 0
    while a + b <= maxValue:
        a, b = b, a + b
        yield b

def test_fib():
    for i in FibIterator():
        print(i, end=" ")
    print()
    for i in fib_generator():
        print(i, end=" ")
    print()
    print(list(FibIterator(maxValue=200)))

test_fib()
````

# 연산자 오버로딩
- 연산자를 메소드로 정의하는 것을 연산자 오버로딩(operator overloading)이라고 한다.
- `a + b`는 내부적으로 `a.__add__(b)`를 호출한다. 클래스 안에 이 메소드를 정의하면 그 클래스의 객체에 `+`를 쓸 수 있다.

| 연산자 | 메소드 | | 연산자 | 메소드 |
|---|---|---|---|---|
| `x + y` | `__add__(self, y)` | | `x == y` | `__eq__(self, y)` |
| `x - y` | `__sub__(self, y)` | | `x != y` | `__ne__(self, y)` |
| `x * y` | `__mul__(self, y)` | | `x > y` | `__gt__(self, y)` |
| `x / y` | `__truediv__(self, y)` | | `x >= y` | `__ge__(self, y)` |
| `x // y` | `__floordiv__(self, y)` | | `x < y` | `__lt__(self, y)` |
| `x % y` | `__mod__(self, y)` | | `x <= y` | `__le__(self, y)` |
| `x ** y` | `__pow__(self, y)` | | `len(x)` | `__len__(self)` |
| `x[i]` | `__getitem__(self, i)` | | `str(x)` | `__str__(self)` |

````python
s1 = "Impossible "
s2 = "Dream"
print(s1.__add__(s2))               # s1 + s2 와 같다.

class Point:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

p1 = Point(1, 2)
p2 = Point(3, 4)
try:
    p1 + p2                         # __add__()가 없어서 오류
except TypeError as e:
    print("TypeError:", e)

class Point:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def __add__(self, other):
        x = self.x + other.x
        y = self.y + other.y
        return Point(x, y)

    def __str__(self):
        return 'Point(' + str(self.x) + ', ' + str(self.y) + ')'

p1 = Point(1, 2)
p2 = Point(3, 4)
print(p1 + p2)
````

# Lab: Book 클래스
- 책을 나타내는 `Book` 클래스를 작성한다.
- `__add__()`: Book 객체끼리의 `+` 연산. 두 책의 페이지 수를 합쳐서 반환한다.
- `__gt__()`: Book 객체끼리의 `>` 연산. 페이지 수로 비교한다.
- `__len__()`: `len(book)`으로 페이지 수를 얻는다.

````python
class Book:
    def __init__(self, title='', pages=0):
        self.title = title
        self.pages = pages

    def __str__(self):
        return self.title

    def __add__(self, other):       # Book 객체끼리의 + 연산을 정의한다.
        return self.pages + other.pages

    def __gt__(self, other):        # Book 객체끼리의 > 연산을 정의한다.
        return self.pages > other.pages

    def __len__(self):
        return self.pages

def test_book():
    book1 = Book('Magic of Python', 600)
    book2 = Book('Master of Python', 700)
    print(book1, "/", book2)
    print(book1 + book2)
    print(book1 > book2)
    print(book1 < book2)            # __lt__()가 없으면 book2.__gt__(book1)을 대신 호출한다.
    print(len(book1))

test_book()
````

# 모듈
- 모듈(module)이란 함수나 변수 또는 클래스들을 모아 놓은 파일이다. (`.py` 파일 하나가 모듈 하나)
- 프로그램이 길어지면 유지 보수를 쉽게 하기 위해 여러 개의 파일로 분할할 수 있다.

| 사용법 | 의미 |
|---|---|
| `import fibo` | 모듈 전체를 임포트. `fibo.fib()`처럼 모듈 이름을 붙여서 사용 |
| `from fibo import fib` | 모듈의 일부 함수만 임포트. `fib()`로 바로 사용 |

| `import fibo as f` | 모듈의 별칭을 만든다. `f.fib()` |
| `import fold.mod` | 폴더로 나누어 그룹화한 모듈 (`fold/mod.py`) |
| `from fold.mod import func` | 폴더 안 모듈의 함수만 임포트 |

# 유용한 모듈: copy
- 프로그래밍에서 중요한 원칙 하나는 이전에 개발된 코드를 적극적으로 재활용하자는 것이다.
- 리스트 같은 객체를 `b = a`로 복사하면 참조 변수만 하나 더 생길 뿐 실제로 객체가 복사되지 않는다.

| 방법 | 최상위 객체 | 내부 객체 |
|---|---|---|
| `b = a` (참조 복사) | 같은 객체 | 같은 객체 |
| `copy.copy(a)` (얕은 복사, shallow copy) | 새로운 객체 | 같은 객체를 참조 |
| `copy.deepcopy(a)` (깊은 복사, deep copy) | 새로운 객체 | 새로운 객체 |

````python
import copy

original_list = [[1, 2, 3], [4, 5, 6]]

ref = original_list                             # 참조 변수만 복사
print(ref is original_list)                     # True

shallow_copied_list = copy.copy(original_list)
print(shallow_copied_list is original_list)     # False: 객체는 복사되지만
print(shallow_copied_list[0] is original_list[0])   # True: 내부 객체는 참조가 유지됨

deep_copied_list = copy.deepcopy(original_list)
print(deep_copied_list[0] is original_list[0])  # False: 객체 및 내부 객체 모두가 복사됨

original_list[0][0] = 100                       # 원본의 내부 객체를 변경하면
print(shallow_copied_list)                      # 얕은 복사본은 같이 바뀌고
print(deep_copied_list)                         # 깊은 복사본은 그대로이다.
````

# 유용한 모듈: keyword

- `keyword.kwlist`: 키워드의 전체 리스트

````python
import keyword

def check_name(name):
    if keyword.iskeyword(name):
        print(name, "은 예약어임.")
    else:
        print(name, "은 사용할 수 있는 변수이름임.")

def test_check_name():
    check_name("for")               # name = input("변수 이름을 입력하시오: ")
    check_name("count")
    print("아래는 키워드의 전체 리스트임: ")
    print(keyword.kwlist)

test_check_name()
````

# Lab: 동전 던지기 게임
- `random` 모듈의 `choice()`로 `"head"`와 `"tail"` 중 하나를 고른다.
- 원래는 `input()`으로 계속할지 물어보면서 반복한다. 여기서는 던질 횟수를 인수로 받는다.

```python
while True:
    response = input("동전 던지기를 계속하시겠습니까?( yes, no) ")
    if response == "yes":
        print(random.choice(myList))
    else:
        break
```

````python
def toss_coins(count):
    myList = ["head", "tail"]
    result = []
    for i in range(count):
        result.append(random.choice(myList))
    return result

def test_toss_coins():
    result = toss_coins(10)
    print(result)
    print("head:", result.count("head"), "tail:", result.count("tail"))

test_toss_coins()
````
