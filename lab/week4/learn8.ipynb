# counter 클래스 만들기
- 데이터: 현재 숫자
- 동작: 증가하기, 현재 값 읽어오기

```python
class Counter:
    def __init__(self):
        self.count = 0

    def increment(self):
        self.count += 1

    def get_count(self):
        return self.count

def test_counter():
    myc = Counter()
    print(myc.get_count())
    
    myc.increment()
    print(myc.get_count())

    myc.increment()
    print(myc.get_count())

    for _ in range(10):
        myc.increment()
    print(myc.get_count())


test_counter()
```
