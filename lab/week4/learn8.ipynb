# counter 클래스 만들기
- 데이터: 현재 숫자
- 동작: 증가하기, 현재 값 읽어오기

```python
class Counter:
    def __init__(self, initial_count=0):
        self.count = initial_count

    def increment(self, amount = 1):
        self.count += amount

    def get_count(self):
        return self.count

def test_counter():
    myc = Counter()
    print(myc.get_count())
    
    myc.increment()
    print(myc.get_count())

    myc.increment()
    print(myc.get_count())

    for _ in range(10):
        myc.increment()
    print(myc.get_count())

    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment()
    print(yourc.get_count())
    yourc.increment(4)
    print(yourc.get_count())


if __name__ == "__main__":
    test_counter()
```

# TV 클래스 추가
- 데이터: 현재 채널번호, 불륨량, ON여부
- 동작: 켜기, 끄기, 채널변경, 불륨변경, 현재 TV 객체 내용 출력

- 클래스명: MyTelevision

```python
class MyTelevision:

    serialNumber = 0
    def __init__(self, channel = 1, volume = 10, on = False):
        self.channel = channel
        self.volume = volume
        self.on = on
        MyTelevision.serialNumber += 1
        self.serialNumber = MyTelevision.serialNumber

    def turnOn(self):
        self.on = True
    def turnOff(self):
        self.on = False
    def changeChannel(self, channel):
        self.channel = channel
    def changeVolume(self, volume):
        self.volume = volume
    def getStatus(self):
        return f"TV가 {'켜져 있습니다.' if self.on else '꺼져 있습니다.'}, 채널: {self.channel}, 볼륨: {self.volume}"

def test_my_television():
    MyTV = MyTelevision(channel = 11)
    print(MyTV.getStatus())
    MyTV.turnOn()
    print(MyTV.getStatus())
    MyTV.turnOff()
    print(MyTV.getStatus())

if __name__ == "__main__":
    test_my_television()
```

# 원 클래스 만들기
- 정보은닉: radius

```python
import math

# Circle 클래스를 정의한다.
class Circle:
    def __init__(self, __radius = 0):
        self.__radius = __radius

    def getArea(self):
        return math.pi * (self.__radius ** 2)

    def getPerimeter(self):
        return 2 * math.pi * self.__radius

# Circle 객체를 생성한다. 
c = Circle(10)
print("원의 면적", c.getArea())
print("원의 둘레", c.getPerimeter())
```

# 사람 클래스 만들기
- 데이터: 출생연도(birth_year: int), 이름(name: str)
- 동작: 올해 몇살인지(get_age(): int), 이름이 어떻게 되는지(get_name() : str)

- class: Person

테스트함수
- 우리 팀의 4명에 대한 객체를 생성하고 목록에 저장: (list 사용)
- for 문을 통해 목록에 있는 모든 객체의 이름과 나이 출력

메인함수
- 테스트 함수 수행

```python
import datetime

class Person:
    def __init__(self, name, year_of_birth):
        self.__name = name
        self.__year_of_birth = year_of_birth

    def get_name(self):
        return self.__name

    def get_age(self, current_year):
        return current_year - self.__year_of_birth

    def get_current_year(self):
        return datetime.datetime.now().year

    def __str__(self):
        return f"이름: {self.get_name()}, 나이: {self.get_age(self.get_current_year())}"

def test_person():
    team_members = [
        Person("이서준", 2007),
        Person("이우성", 2007),
        Person("신재중", 2007),
        Person("이찬민", 2007)
    ]
    for member in team_members:
        print(member)

if __name__ == "__main__":
    test_person()
```

# 자동차 클래스 작성
- 속성: 메이커, 모델, 색상, 연식, 가격
- 동작: 주행하기, 방향바꾸기, 주차하기

```python
class Car:
    def __init__(self, speed, color, model):
        self.speed = speed
        self.color = color
        self.model = model

    def drive(self):
        self.speed = 60

def test_car():
    print("자동차 객체를 생성하였습니다.")
    myCar = Car(0, "Red", "Sedan")
    print("자동차의 속도는", myCar.speed)
    print("자동차의 색상은", myCar.color)
    print("자동차의 모델은", myCar.model)

    myCar.drive()
    print("자동차의 속도는", myCar.speed)

if __name__ == "__main__":
    test_car()
```

# 은행 계좌 클래스

```python
class BankAccount:
    def __init__(self):
        self.__balance = 0

    def withdraw(self, amount):
        self.__balance -= amount
        print(f"통장에 {amount}가 입금되었음.")
        return self.__balance

    def deposit(self, amount):
        self.__balance += amount
        print(f"통장에 {amount}가 출금되었음.")
        return self.__balance

def test_bank_account():
    account = BankAccount()
    print("현재 잔액:", account.deposit(1000))
    print("현재 잔액:", account.withdraw(500))
    print("현재 잔액:", account.deposit(2000))
    print("현재 잔액:", account.withdraw(1500))

if __name__ == "__main__":
    test_bank_account()
```

# 벡터 클래스 모델링

```python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector2D(self.x - other.x, self.y - other.y) 

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __str__(self):
        return '(%g, %g)' % (self.x, self.y)

def test_vector2d():
    u = Vector2D(0, 1)
    v = Vector2D(1, 0)
    w = Vector2D(1, 1)
    a = u + v
    print(a)

test_vector2d()
```

```python
# programming 3
class Box:
    def __init__(self, l, h, d):
        self.length = l
        self.height = h
        self.depth = d
    def __str__(self):
        return "(%s, %s, %s)" % (self.length, self.height, self.depth)
    def setLength(self, l): self.length = l
    def getLength(self): return self.length
    def setHeight(self, h): self.height = h
    def getHeight(self): return self.height
    def setDepth(self, d): self.depth = d
    def getDepth(self): return self.depth
```

```python
# programming 4

class Rectangle:
    def __init__(self, x, y, w, h):
        self.x = x; self.y = y
        self.width = w; self.height = h
    def __str__(self):
        return "(%s, %s, %s, %s)" % (self.x, self.y, self.width, self.height)
    def setX(self, x): self.x = x
    def getX(self): return self.x
    def getArea(self):
        return self.width * self.height
    def overlap(self, r):
        return (self.x < r.x + r.width and r.x < self.x + self.width and
                self.y < r.y + r.height and r.y < self.y + self.height)

r1 = Rectangle(0, 0, 100, 100)
r2 = Rectangle(10, 10, 100, 100)
if r1.overlap(r2):
    print("r1과 r2는 서로 겹칩니다.")
```

```python
# programming 5

class Triangle:
    def __init__(self, a1, a2, a3):
        self.angle1 = a1
        self.angle2 = a2
        self.angle3 = a3
        self.numberOfSides = 3
    def __str__(self):
        return "(%s, %s, %s)" % (self.angle1, self.angle2, self.angle3)
    def setAngle1(self, a): self.angle1 = a
    def getAngle1(self): return self.angle1
    # angle2, angle3도 같은 방식으로
    def checkAngles(self):
        return self.angle1 + self.angle2 + self.angle3 == 180
```

```python
# programming 7

class PhoneBook:
    def __init__(self):
        self.contacts = {}
    def add(self, name, mobile=None, office=None, email=None):
        self.contacts[name] = Person(name, mobile, office, email)
    def __str__(self):
        return "\n\n".join(str(p) for p in self.contacts.values())
```
