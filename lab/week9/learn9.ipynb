# tkinter란?
- 파이썬에서 그래픽 사용자 인터페이스(GUI: Graphical User Interface)를 개발할 때 사용하는 표준 모듈
- 유닉스 계열에서 쓰이던 **Tcl/Tk** 위에 객체 지향 계층을 입힌 것이다.
  - Tk는 John Ousterhout가 Tcl 스크립팅 언어를 위한 GUI 확장으로 개발
- 텍스트 모드: `input()`/`print()`로 순서대로 입출력
- GUI 모드: 윈도우에 버튼, 입력창 등을 배치하고 **사용자의 동작(이벤트)을 기다렸다가** 처리

> 이 노트북의 코드 셀은 실행하면 별도의 윈도우가 뜬다. 윈도우를 닫아야 셀 실행이 끝나고 다음 셀을 실행할 수 있다.

# tkinter 시작하기
1. 루트 윈도우(window)를 생성한다.
2. 위젯을 생성할 때 **첫 번째 인수로 부모 윈도우**를 넘긴다.
3. 배치 관리자(`pack()`, `grid()`, `place()`)로 위젯을 윈도우에 배치한다.
4. `mainloop()`로 이벤트 처리 루프에 들어간다.

위젯의 종류
- 단순 위젯: `Button`, `Canvas`, `Checkbutton`, `Entry`, `Label`, `Message` 등
- 컨테이너 위젯: 다른 위젯을 안에 포함할 수 있는 위젯. `Frame`, `Toplevel`, `LabelFrame`, `PanedWindow` 등

````python
from tkinter import *    # tkinter 모듈을 포함

window = Tk()                                # 루트 윈도우를 생성
label = Label(window, text="Hello tkinter")  # 레이블 위젯을 생성
label.pack()                                 # 레이블 위젯을 윈도우에 배치

window.mainloop()                            # 윈도우가 사용자 동작을 대기
````

분석
- `from tkinter import *` : tkinter 모듈을 포함시킨다.
- `window = Tk()` : Tk 클래스의 객체가 생성되면서 화면에 하나의 윈도우가 생성된다.
- `label = Label(window, text="Hello tkinter")` : `window`를 부모로 하는 레이블 위젯을 생성한다.
- `label.pack()` : 압축(pack) 배치 관리자를 이용하여 레이블을 컨테이너에 위치시킨다. (`pack()`을 안 하면 화면에 안 보인다.)
- `window.mainloop()` : 이벤트 처리 루프. 사용자로부터 오는 마우스나 키보드 이벤트를 처리한다. 윈도우를 닫으면 끝난다.

# tkinter의 위젯들
| 위젯 | 설명 |
|---|---|
| `Button` | 클릭하면 동작하는 버튼 |
| `Label` | 텍스트나 이미지를 표시 |
| `Entry` | 한 줄 텍스트 입력 |
| `Text` | 여러 줄 텍스트 입력 |
| `Canvas` | 도형, 이미지를 그리는 영역 |
| `Checkbutton` / `Radiobutton` | 체크 박스 / 라디오 버튼 |
| `Scale` | 슬라이더로 값 선택 |
| `Frame` | 다른 위젯을 담는 컨테이너 |

공통으로 자주 쓰는 옵션
- `text` : 표시할 문자열
- `bg`(background) : 배경색, `fg`(foreground) : 전경색(글자색)
- `width`, `height` : 크기 (버튼/레이블은 **글자 수, 줄 수** 단위)
- `font` : 폰트 (예: `("Helvetica", 16)`)

# 배치 관리자
위젯을 윈도우 안에 어떻게 배치할지 결정한다.

| 배치 관리자 | 메소드 | 특징 |
|---|---|---|
| 압축(pack) 배치 관리자 | `pack()` | 위젯을 위에서 아래로(또는 `side`방향으로) 차곡차곡 쌓는다 |
| 격자(grid) 배치 관리자 | `grid()` | 위젯을 행(row)과 열(column)의 테이블 형태로 배치 |
| 절대(place) 배치 관리자 | `place()` | 위젯을 지정한 절대 좌표(x, y)에 배치 |

> 하나의 컨테이너 안에서는 `pack()`과 `grid()`를 섞어 쓰면 안 된다. 섞고 싶으면 `Frame`으로 나눈다.

# 압축(pack) 배치 관리자
- 기본은 위에서 아래로 쌓인다.

````python
from tkinter import *       # 교재의 Tkinter는 파이썬 2 표기. 파이썬 3에서는 소문자 tkinter
window = Tk()
Label(window, text="박스 #1", bg="red", fg="white").pack()
Label(window, text="박스 #2", bg="green", fg="black").pack()
Label(window, text="박스 #3", bg="blue", fg="white").pack()
window.mainloop()
````

- `side=LEFT` 옵션을 주면 왼쪽부터 가로로 쌓인다. (`TOP`, `BOTTOM`, `LEFT`, `RIGHT`)

````python
from tkinter import *
window = Tk()
Button(window, text="박스 #1", bg="red", fg="white").pack(side=LEFT)
Button(window, text="박스 #2", bg="green", fg="black").pack(side=LEFT)
Button(window, text="박스 #3", bg="orange", fg="white").pack(side=LEFT)
window.mainloop()
````

# 격자(grid) 배치 관리자
- 위젯(버튼, 레이블 등)을 테이블 형태로 배치한다.
- `grid(row=행, column=열)` : 0부터 시작
- `columnspan`, `rowspan` : 여러 열/행을 합쳐서 차지
- `sticky` : 셀 안에서 붙을 방향 (`N`, `S`, `E`, `W`, 예: `sticky=W+E`는 가로로 꽉 채움)

````python
from tkinter import *
window = Tk()

b1 = Button(window, text="박스 #1", bg="red", fg="white")
b2 = Button(window, text="박스 #2", bg="green", fg="white")
b3 = Button(window, text="박스 #3", bg="orange", fg="white")
b4 = Button(window, text="박스 #4", bg="pink", fg="white")

b1.grid(row=0, column=0)  # 0행 0열
b2.grid(row=0, column=1)  # 0행 1열
b3.grid(row=1, column=0)  # 1행 0열
b4.grid(row=1, column=1)  # 1행 1열

window.mainloop()
````

# 절대(place) 배치 관리자
- `place(x=, y=)` : 윈도우의 왼쪽 위가 (0, 0)이고, 픽셀 단위로 위치를 지정한다.

````python
from tkinter import *
window = Tk()
b1 = Button(window, text="박스 #1", bg="red", fg="white")
b1.place(x=0, y=0)
b2 = Button(window, text="박스 #2", bg="green", fg="black")
b2.place(x=20, y=30)
b3 = Button(window, text="박스 #3", bg="orange", fg="white")
b3.place(x=40, y=60)
window.mainloop()
````

# 여러 배치 관리자 혼용하기
- `Frame`(컨테이너 위젯) 안에 버튼들을 가로로 배치하고, 그 프레임을 다시 윈도우에 배치한다.
- `pack()`을 호출한 **순서대로** 쌓이므로 레이블이 프레임보다 위에 나온다.

````python
from tkinter import *

window = Tk()
f = Frame(window)

b1 = Button(f, text="박스 #1", bg="red", fg="white")
b2 = Button(f, text="박스 #2", bg="green", fg="black")
b3 = Button(f, text="박스 #3", bg="orange", fg="white")
b1.pack(side=LEFT)
b2.pack(side=LEFT)
b3.pack(side=LEFT)

l = Label(window, text="이 레이블은 버튼들 위에 배치된다.")
l.pack()
f.pack()

window.mainloop()
````

# 버튼 이벤트 처리
- 이벤트 구동 프로그래밍(event-driven programming): 프로그램이 순서대로 실행되는 것이 아니라, 이벤트가 발생하면 그에 연결된 함수가 호출된다.
- 버튼을 클릭했을 때 호출될 함수를 **콜백(callback) 함수** 또는 이벤트 처리 함수라고 한다.
- `Button(..., command=함수이름)` : 함수 이름만 넘긴다. `command=process()`처럼 괄호를 붙이면 생성 시점에 바로 호출되어 버린다.

````python
from tkinter import *

def process():
    print("버튼이 클릭되었습니다.")

window = Tk()
button = Button(window, text="클릭하세요!", command=process)
button.pack()

window.mainloop()
````

# Lab: 카운터 만들기
- 증가 버튼을 누르면 카운터가 1 증가된다.
- 함수 안에서 전역 변수 `counter`의 값을 바꾸려면 `global` 선언이 필요하다.
- 위젯의 옵션은 `label['text'] = ...` 또는 `label.config(text=...)`로 실행 중에 바꿀 수 있다.

````python
from tkinter import *
window = Tk()

counter = 0
def clicked():
    global counter
    counter += 1
    label['text'] = '버튼 클릭 횟수: ' + str(counter)

label = Label(window, text="아직 눌려지지 않음")
label.pack()
Button(window, text="증가", command=clicked).pack()
window.mainloop()
````

# Lab: 온도 변환기 #2
- "화씨->섭씨" 버튼을 누르면 입력한 화씨 온도가 섭씨 온도로 변환되어 나타나게 한다.
- 섭씨 = (화씨 - 32) × 5 / 9

````python
from tkinter import *

# 이벤트 처리 함수를 정의한다.
def process():
    tf = float(e1.get())        # e1에서 문자열을 읽어서 부동소수점형으로 변경
    tc = (tf - 32.0) * 5.0 / 9.0  # 화씨 온도를 섭씨 온도로 변환한다.
    e2.delete(0, END)           # 처음부터 끝까지 지운다.
    e2.insert(0, str(tc))       # tc 변수의 값을 문자열로 변환하여 추가한다.

window = Tk()

Label(window, text="화씨").grid(row=0, column=0)
Label(window, text="섭씨").grid(row=1, column=0)

e1 = Entry(window)
e2 = Entry(window)
e1.grid(row=0, column=1)
e2.grid(row=1, column=1)

Button(window, text="화씨->섭씨", command=process).grid(row=2, column=1)
window.mainloop()
````

# Lab: 숫자 추측 게임
- 컴퓨터가 생성한 1~100 사이의 난수를 GUI로 알아맞히는 게임
- "시도" 버튼: 입력값과 정답을 비교해 "높음!", "낮음!", "정답!" 출력
- "초기화" 버튼: 정답을 다시 설정
- `window.configure(bg=...)`로 윈도우 배경색을, `window.title()`로 제목을 설정한다.

````python
from tkinter import *
import random

answer = random.randint(1, 100)   # 정답을 1에서 100 사이의 난수로 설정한다.

def guessing():
    guess = int(guessField.get())   # 텍스트 필드에서 사용자가 입력한 값을 가져온다.
    if guess > answer:
        msg = "높음!"
    elif guess < answer:
        msg = "낮음!"
    else:
        msg = "정답!"
    resultLabel["text"] = msg       # 메시지를 출력한다.
    guessField.delete(0, END)

def reset():                        # 정답을 다시 설정한다.
    global answer
    answer = random.randint(1, 100)
    resultLabel["text"] = "다시 한번 하세요!"

window = Tk()
window.configure(bg="white")
window.title("숫자를 맞춰보세요!")
window.geometry("500x80")

titleLabel = Label(window, text="숫자 게임에 오신 것을 환영합니다!", bg="white")
titleLabel.pack()

guessField = Entry(window)
guessField.pack(side="left")

tryButton = Button(window, text="시도", fg="green", bg="white", command=guessing)
tryButton.pack(side="left")

resetButton = Button(window, text="초기화", fg="red", bg="white", command=reset)
resetButton.pack(side="left")

resultLabel = Label(window, text="1부터 100사이의 숫자를 입력하시오.", bg="white")
resultLabel.pack(side="left")

window.mainloop()
````

# Lab: 가위, 바위, 보 게임
- 사용자가 버튼으로 가위/바위/보를 선택하면 컴퓨터가 무작위로 선택하고 승패를 출력한다.
- 교재는 `rock.png`, `paper.png`, `scissors.png` 이미지를 버튼에 넣지만(`Button(frame, image=rock_image, ...)`), 여기서는 이미지 파일 없이 실행되도록 텍스트로 표시했다.
- `command`에는 인수 없는 함수만 넘길 수 있으므로 `pass_s()`, `pass_r()`, `pass_p()` 같은 중간 함수를 만들어 `decide()`를 호출한다. (`lambda: decide("가위")`로도 가능)
- `output.config(text=...)` : 위젯의 옵션 변경

````python
import random
from tkinter import *

window = Tk()
Label(window, text="선택하세요", font=("Helvetica", "16")).pack()
frame = Frame(window)

# 이미지를 사용할 경우
# rock_image = PhotoImage(file="rock.png")
# paper_image = PhotoImage(file="paper.png")
# scissors_image = PhotoImage(file="scissors.png")

def pass_s():
    decide("가위")

def pass_r():
    decide("바위")

def pass_p():
    decide("보")

def decide(human):
    computer = random.choice(["가위", "바위", "보"])
    computer_label["text"] = computer      # 이미지 사용 시: computer_image["image"] = rock_image 등

    if (computer == "바위" and human == "보") or (computer == "보" and human == "가위") \
            or (computer == "가위" and human == "바위"):
        result = "인간 승리!"
    elif computer == human:
        result = "비겼습니다."
    else:
        result = "컴퓨터 승리!"
    output.config(text="인간: " + human + " 컴퓨터: " + computer + " " + result)

rock = Button(frame, text="바위", width=8, command=pass_r)
rock.pack(side="left")
paper = Button(frame, text="보", width=8, command=pass_p)
paper.pack(side="left")
scissors = Button(frame, text="가위", width=8, command=pass_s)
scissors.pack(side="left")
frame.pack()

Label(window, text="컴퓨터는 다음을 선택하였습니다.", font=("Helvetica", "16")).pack()
computer_label = Label(window, text="?", font=("Helvetica", "30"))
computer_label.pack()
output = Label(window, text="", font=("Helvetica", "16"))
output.pack()

window.mainloop()
````

# Lab: TIC-TAC-TOE 게임
- 3×3칸의 게임판에서 두 경기자가 번갈아 O와 X를 표시한다.
- 버튼 9개를 반복문으로 만들어 `grid(row=i//3, column=i%3)`로 배치하고 리스트에 저장한다.
- `command=lambda k=i: checked(k)` : 버튼마다 자기 번호를 인수로 넘기기 위해 람다의 **기본값**을 사용한다.
  - `lambda: checked(i)`로 쓰면 반복문이 끝난 뒤의 `i`(=8)를 모든 버튼이 공유하게 된다.
- 교재는 리스트 변수 이름을 `list`로 썼지만 내장 함수 `list()`를 가리게 되므로 `buttons`로 바꿨다.

````python
from tkinter import *

# i번째 버튼을 누를 수 있는지 검사한다. 누를 수 있으면 X나 O를 표시한다.
def checked(i):
    global player
    button = buttons[i]         # 리스트에서 i번째 버튼 객체를 가져온다.

    # 버튼이 초기상태가 아니면 이미 누른 버튼이므로 아무것도 하지 않고 리턴한다.
    if button["text"] != "     ":
        return
    button["text"] = "  " + player + "  "
    if player == "X":
        player = "O"
        button["bg"] = "yellow"
    else:
        player = "X"
        button["bg"] = "lightgreen"

window = Tk()               # 윈도우를 생성한다.
player = "X"                # 시작은 플레이어 X이다.
buttons = []

# 9개의 버튼을 생성하여 격자 형태로 윈도우에 배치한다.
for i in range(9):
    b = Button(window, text="     ", command=lambda k=i: checked(k))
    b.grid(row=i // 3, column=i % 3)
    buttons.append(b)       # 버튼 객체를 리스트에 저장한다.

window.mainloop()
````

# Lab: 계산기 프로그램
- 격자 배치 관리자와 버튼, 엔트리 위젯만으로 계산기를 만든다.
- 엔트리(표시창)는 0행에서 `columnspan=5`로 5열을 모두 차지한다.
- 버튼 20개는 리스트로 만들고, 반복문으로 1행부터 5개씩 배치한다.
- 교재는 화면 배치까지만 있으므로, 버튼을 누르면 동작하도록 `click()` 함수를 추가했다.
  - 숫자/연산자 버튼: 표시창 끝에 문자를 추가
  - `=` : `eval()`로 수식을 계산 (`eval("3+4*2")` → `11`)
  - `C` : 표시창을 지움

````python
from tkinter import *

def click(key):
    if key == "=":
        try:
            result = eval(display.get())    # 문자열 수식을 계산한다.
            display.delete(0, END)
            display.insert(END, str(result))
        except Exception:
            display.delete(0, END)
            display.insert(END, "오류")
    elif key == "C":
        display.delete(0, END)
    else:
        display.insert(END, key)            # 끝에 문자를 추가한다.

window = Tk()
window.title("My Calculator")
display = Entry(window, width=33, bg="yellow")
display.grid(row=0, column=0, columnspan=5)

button_list = [
    '7', '8', '9', '/', 'C',
    '4', '5', '6', '*', ' ',
    '1', '2', '3', '-', ' ',
    '0', '.', '=', '+', ' ']

row_index = 1
col_index = 0
for button_text in button_list:
    Button(window, text=button_text, width=5,
           command=lambda t=button_text: click(t)).grid(row=row_index, column=col_index)
    col_index += 1
    if col_index > 4:
        row_index += 1
        col_index = 0

window.mainloop()
````

# Lab: 그림판 프로그램 만들기
- 마우스를 드래그하면 캔버스에 선이 그려지는 그림판
- 버튼: 펜, 브러쉬, 색상선택, 지우개, 모두삭제 / `Scale`로 선 굵기 조절
- 동작 원리
  - `<B1-Motion>` : 드래그하는 동안 `paint()`가 계속 호출된다. 이전 좌표(`old_x`, `old_y`)에서 현재 좌표까지 선을 그린다.
  - `<ButtonRelease-1>` : 버튼을 떼면 `reset()`으로 이전 좌표를 `None`으로 지워서, 다음에 그릴 때 떨어진 선과 이어지지 않게 한다.
  - 지우개는 배경색(흰색)으로 그리는 것이다.
- `askcolor()` : `tkinter.colorchooser` 모듈의 색상 선택 대화상자. `(RGB튜플, "#16진수")`를 반환하므로 `[1]`을 사용한다.
- `DoubleVar()` : 위젯과 연결되는 tkinter 변수. `Scale(variable=var)`로 연결하면 `var.get()`으로 현재 값을 읽을 수 있다.
- 교재는 일부 코드만 있으므로 전체 프로그램으로 완성했다. (교재의 `use_erase`/`use_eraser` 이름 불일치도 통일)

````python
from tkinter import *
from tkinter.colorchooser import askcolor

mode = "pen"
mycolor = "black"
old_x, old_y = None, None

def use_pen():          # 펜 버튼이 선택되면 모드를 "pen"으로 바꾼다.
    global mode
    mode = "pen"

def use_brush():
    global mode
    mode = "brush"

def choose_color():
    global mycolor
    color = askcolor(color=mycolor)[1]
    if color:                       # 취소를 누르면 None이 반환된다.
        mycolor = color

def use_eraser():
    global mode
    mode = "erase"

def clear_all():
    canvas.delete(ALL)

def paint(event):
    global old_x, old_y
    fill_color = "white" if mode == "erase" else mycolor
    width = var.get() * 3 if mode == "brush" else var.get()   # 브러쉬는 더 굵게
    if old_x and old_y:
        canvas.create_line(old_x, old_y, event.x, event.y,
                           capstyle=ROUND, width=width, fill=fill_color)
    old_x = event.x
    old_y = event.y

def reset(event):
    global old_x, old_y
    old_x, old_y = None, None

window = Tk()
window.title("그림판")

penButton = Button(window, text='펜', command=use_pen)
penButton.grid(row=0, column=0, sticky=W+E)

brushButton = Button(window, text='브러쉬', command=use_brush)
brushButton.grid(row=0, column=1, sticky=W+E)

colorButton = Button(window, text='색상선택', command=choose_color)
colorButton.grid(row=0, column=2, sticky=W+E)

eraseButton = Button(window, text='지우개', command=use_eraser)
eraseButton.grid(row=0, column=3, sticky=W+E)

clearButton = Button(window, text='모두삭제', command=clear_all)
clearButton.grid(row=0, column=4, sticky=W+E)

var = DoubleVar(value=1)
scale = Scale(window, variable=var, from_=1, to=20, orient=VERTICAL)
scale.grid(row=1, column=5, sticky=N+S)

canvas = Canvas(window, bg='white', width=600, height=400)
canvas.grid(row=1, columnspan=5)

canvas.bind('<B1-Motion>', paint)
canvas.bind('<ButtonRelease-1>', reset)

window.mainloop()
````

# Lab: 공 애니메이션 I
- 크기와 색상이 다른 3개의 공이 캔버스 벽에 부딪히면 반사된다.
- `Ball` 클래스
  - `__init__` : 캔버스에 원을 그리고 id를 저장, 이동 속도 `dx`, `dy`를 난수로 정한다.
  - `move()` : `canvas.move()`로 이동 후 `canvas.coords()`로 현재 좌표를 얻어, 벽을 벗어나면 속도의 부호를 반전시킨다.
- `canvas.winfo_width()`, `canvas.winfo_height()` : 캔버스의 실제 크기
- 애니메이션 루프: `move()` → `window.update()`(화면 갱신) → `time.sleep(0.05)`(잠시 대기)를 반복한다.
  - `while True` 루프 중에 윈도우를 닫으면 이미 없어진 위젯에 접근하므로 `TclError`가 난다. 그래서 `try`~`except`로 루프를 빠져나오게 했다.

````python
from tkinter import *
import time
import random

window = Tk()
canvas = Canvas(window, width=600, height=400)
canvas.pack()

class Ball():
    def __init__(self, color, size):
        self.id = canvas.create_oval(0, 0, size, size, fill=color)
        self.dx = random.randint(1, 10)
        self.dy = random.randint(1, 10)

    def move(self):
        canvas.move(self.id, self.dx, self.dy)
        x0, y0, x1, y1 = canvas.coords(self.id)
        if y1 > canvas.winfo_height() or y0 < 0:    # 원이 위쪽이나 아래쪽으로 벗어났으면
            self.dy = -self.dy                      # dy의 부호를 반전시킨다.
        if x1 > canvas.winfo_width() or x0 < 0:     # 원이 왼쪽이나 오른쪽으로 벗어났으면
            self.dx = -self.dx                      # dx의 부호를 반전시킨다.

ball1 = Ball("blue", 60)
ball2 = Ball("green", 100)
ball3 = Ball("orange", 80)

try:
    while True:
        ball1.move()
        ball2.move()
        ball3.move()
        window.update()
        time.sleep(0.05)
except TclError:            # 윈도우를 닫으면 루프를 끝낸다.
    pass
````

# Lab: 공 애니메이션 II
- 공을 30개 정도 움직이려면 개별 변수로 각 객체를 참조하는 것은 거의 불가능하다.
- 이럴 때는 **리스트에 객체를 저장**하고 반복문으로 처리한다.
- 아래는 `after()`를 사용한 방법이다. `window.after(밀리초, 함수)`는 지정한 시간 후에 함수를 한 번 호출한다. 함수 안에서 다시 `after()`를 호출하면 `mainloop()`를 막지 않고 애니메이션을 반복할 수 있다. (`while`+`sleep` 방식보다 tkinter에 적합)

````python
from tkinter import *
import random

window = Tk()
canvas = Canvas(window, width=600, height=400)
canvas.pack()

class Ball():
    def __init__(self, color, size):
        self.id = canvas.create_oval(0, 0, size, size, fill=color)
        self.dx = random.randint(1, 10)
        self.dy = random.randint(1, 10)

    def move(self):
        canvas.move(self.id, self.dx, self.dy)
        x0, y0, x1, y1 = canvas.coords(self.id)
        if y1 > canvas.winfo_height() or y0 < 0:
            self.dy = -self.dy
        if x1 > canvas.winfo_width() or x0 < 0:
            self.dx = -self.dx

colors = ["red", "orange", "yellow", "green", "blue", "indigo", "violet"]
ballList = []
for i in range(30):
    ballList.append(Ball(random.choice(colors), 60))

def animate():
    for ball in ballList:
        ball.move()
    window.after(50, animate)   # 50ms(0.05초) 후에 다시 호출

animate()
window.mainloop()
````

# 이번 장에서 배운 것
- tkinter에서는 먼저 루트 윈도우(`Tk()`)를 생성하고, 레이블이나 버튼을 생성할 때 첫 번째 인수로 윈도우를 넘기면 된다. 마지막에 `mainloop()`를 호출한다.
- 파이썬은 3종류의 배치 관리자를 제공한다: 압축(pack) 배치 관리자, 격자(grid) 배치 관리자, 절대(place) 배치 관리자
- 버튼을 클릭했을 때 실행할 함수는 `command=함수이름`으로 지정한다.
- 위젯에 이벤트를 처리하는 함수를 연결하려면 `bind()` 메소드를 사용한다. 예: `widget.bind('<Button-1>', callback)`
- `Canvas`에 `create_line`, `create_rectangle`, `create_oval` 등으로 도형을 그리고, 반환된 id로 `move`, `coords`, `itemconfig`, `delete` 할 수 있다.
- 애니메이션은 도형을 조금씩 움직이고 화면을 갱신하는 것을 반복하여 만든다. (`update()`+`sleep()` 또는 `after()`)
