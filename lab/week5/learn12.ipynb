## 상속

- 부모 클래스를 상속받아서 자식 클래스를 정의할 수 있다.
- 부모 클래스의 메소드를 자식 클래스에서 재정의(오버라이딩)할 수 있다.
- object 클래스를 이해할 수 있다.
- 클래스 간의 관계(is-a, has-a)를 파악할 수 있다.

# 객체지향의 3가지 핵심요소
- 캡슐화: 데이터와 연산을 하나의 덩어리로 묶어서 다룸
  - 정보은닉을 위한 가시성: public(공개), protected(보호), private(숨김)
- 상속: 부모 클래스를 자식 클래스가 그대로 물려받아 추가적인 코드를 작성할 수 있음
  - 부모의 private 한 것은 상속은 받으나 접근 불가능
  - protected는 외부 접근은 권장되지 않지만 자식이 물려받아 사용 가능
- 다형성: 함수 이름이 같더라도 문맥(클래스)에 따라 다른 함수가 수행됨
  - A.do(), B.do() 처럼 이름은 같으나 클래스가 다르면 다른 함수
  - B가 A를 상속하고 재정의할 경우 동적 바인딩이 가능

___
상속(inheritance)은 기존에 존재하는 클래스로부터 코드와 데이터를 이어받고 자신이 필요한 기능을 추가하는 기법이다.

상속은 클래스 간의 "is-a" 관계를 만든다.
- 푸들은 강아지이다.
- 자동차는 차량이다.
- 사각형은 모양이다.

부모 클래스 = 슈퍼 클래스(super-class), 자식 클래스 = 서브 클래스(sub-class)

문법: `class 자식클래스(부모클래스):`

상속을 사용하면 중복된 코드를 부모 클래스 한 곳으로 모을 수 있다.

# Car 클래스와 ElectricCar, Truck 클래스
- 부모: Car (메이커, 모델, 색상, 가격)
- 자식: ElectricCar (배터리 크기 추가), Truck (적재량 추가)
- 자식 클래스의 생성자에서 `super().__init__()`으로 부모 클래스의 생성자를 호출
- `__str__()`을 재정의하고, 부모의 `__str__()` 결과에 자식의 정보를 덧붙임

같은 코드가 `Car.py`에도 있음

```python
class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price

    def setMaker(self, maker):
        self.maker = maker

    def getMaker(self):
        return self.maker

    def __str__(self):
        return f"maker: {self.maker}, model: {self.model}, color: {self.color}, price: {self.price}"

    def getDesc(self):
        return self.__str__()

class ElectricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.bat_size = bat_size

    def set_bat_size(self, bat_size):
        self.bat_size = bat_size

    def get_bat_size(self):
        return self.bat_size

    def __str__(self):
        return f"{super().__str__()}, Battery : {self.bat_size}"

class Truck(Car):
    def __init__(self, maker, model, color, price, payLoad):
        super().__init__(maker, model, color, price)
        self.payLoad = payLoad

    def set_payLoad(self, payLoad):
        self.payLoad = payLoad

    def get_payLoad(self):
        return self.payLoad

    def __str__(self):
        return f"{super().__str__()}, PayLoad : {self.payLoad}"

def test_car():
    car1 = ElectricCar("Tisla", "Model S", "white", 100000, 0)
    car1.setMaker("Tesla")      # 부모에게서 물려받은 메소드
    car1.set_bat_size(60)       # 자식에서 추가한 메소드
    print(car1.getDesc())

    car2 = Truck("Kia", "truw", "Blue", 120000, 5000)
    car2.set_payLoad(6000)
    print(car2.getDesc())

if __name__ == "__main__":
    test_car()
```

# 부모 클래스의 생성자 호출
- 자식 클래스에서 `__init__()`을 새로 만들면 부모의 `__init__()`은 자동으로 호출되지 않는다.
- `super().__init__()`을 호출하지 않으면 부모 클래스에서 만드는 속성이 생성되지 않아서 오류가 난다.

```python
class Animal:
    def __init__(self, age=0):
        self.age = age

    def eat(self):
        print("동물이 먹고 있습니다.")

# 부모 클래스의 생성자를 호출하지 않았다!
class BadDog(Animal):
    def __init__(self, age=0, name=""):
        self.name = name

class Dog(Animal):
    def __init__(self, age=0, name=""):
        super().__init__(age)
        self.name = name

def test_parent_init():
    d = BadDog()
    try:
        print(d.age)
    except AttributeError as e:
        print("오류:", e)       # age 변수가 생성되지 않았다.

    d = Dog(3, "바둑이")
    print(d.age)
    d.eat()                     # 메소드는 생성자와 상관없이 상속된다.

if __name__ == "__main__":
    test_parent_init()
```

# type()과 isinstance() 함수
- `type(x)`: 객체를 만든 바로 그 클래스를 알려준다.
- `isinstance(x, 클래스)`: 부모 클래스까지 포함해서 그 클래스의 객체인지 확인한다. (자식 객체는 부모 클래스의 객체이기도 하다.)

```python
x = Animal()
y = Dog()
print(type(x))
print(type(y))
print(isinstance(x, Animal), isinstance(y, Animal))
print(isinstance(x, Dog), isinstance(y, Dog))
print(type(y) == Animal)    # type()은 부모 클래스와는 같지 않다.
```

# 가시성 (Visibility)
파이썬은 C++/자바와 달리 접근 제어자가 엄격하게 적용되지 않고, 이름 앞의 밑줄로 관례적으로 표현한다.

| 가시성 | 표기 | 의미 |
|---|---|---|
| public | `value` | 어디서나 접근 가능 |
| protected | `_value` | 자식 클래스에서 접근 권장, 외부에서 직접 접근은 권장하지 않음 (막지는 않음) |
| private | `__value` | 클래스 내부에서만 접근. 이름이 `_클래스명__value`로 바뀌어(name mangling) 외부/자식에서 접근하기 어려움 |

```python
class Base:
    def __init__(self):
        self.public_value = 1         # public
        self._protected_value = 2     # protected
        self.__private_value = 3      # private

class Child(Base):
    def show(self):
        print(self.public_value)          # 가능
        print(self._protected_value)      # 가능
        try:
            print(self.__private_value)   # 불가 (AttributeError)
        except AttributeError as e:
            print("오류:", e)

def test_visibility():
    c = Child()
    c.show()
    print(c._protected_value)       # 가능 (하지만 권장되지 않음)
    print(c._Base__private_value)   # 실제로는 이 이름으로 저장되어 있다.

if __name__ == "__main__":
    test_visibility()
```

# 다중 상속
- 파이썬은 부모 클래스를 2개 이상 가질 수 있다.
- 부모가 여러 명이므로 `super()`보다는 `부모클래스.__init__(self, ...)`처럼 부모 클래스의 이름으로 접근한다.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
        print(self.name, self.age)

class Student:
    def __init__(self, id):
        self.id = id

    def getId(self):
        return self.id

class CollegeStudent(Person, Student):
    def __init__(self, name, age, id):
        Person.__init__(self, name, age)
        Student.__init__(self, id)

obj = CollegeStudent('Kim', 22, '100036')
obj.show()
print(obj.getId())
```

# 메소드 오버라이딩
- 부모 클래스에서 정의된 메소드를 자식 클래스에서 같은 이름으로 다시 정의하는 것
- "자식 클래스의 메소드가 부모 클래스의 메소드를 오버라이드(재정의)한다"고 말한다.
- 재정의한 메소드 안에서 부모의 메소드도 실행하고 싶으면 `super().메소드()`를 호출한다.

```python
import math

class Shape:
    def __init__(self):
        pass

    def draw(self):
        print("draw()가 호출됨")

    def get_area(self):
        print("get_area()가 호출됨")

class Circle(Shape):
    def __init__(self, radius=0):
        super().__init__()
        self.radius = radius

    def draw(self):
        super().draw()              # 부모 클래스의 draw()가 호출된다.
        print("원을 그립니다.")

    def get_area(self):
        return math.pi * self.radius ** 2

c = Circle(10)
c.draw()
print("원의 면적:", c.get_area())
```

# Lab: 직원과 매니저
- 직원(Employee)은 월급만 있고, 매니저(Manager)는 월급 외에 보너스가 있다.
- `getSalary()`를 Manager에서 재정의: 부모의 `getSalary()` 결과에 보너스를 더한다.
- 강의자료의 `return salary`는 `self`가 빠져서 NameError가 난다 → `return self.salary`

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def getSalary(self):
        return self.salary

class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = bonus

    def getSalary(self):
        salary = super().getSalary()
        return salary + self.bonus

    def __repr__(self):
        return f"이름: {self.name}; 월급: {self.salary}; 보너스: {self.bonus}"

def test_manager():
    kim = Manager("김철수", 2000000, 1000000)
    print(kim)
    print("받는 금액:", kim.getSalary())

if __name__ == "__main__":
    test_manager()
```

# Lab: 은행 계좌
- BankAccount: 잔액(balance), 이름(name), 통장 번호(number), 출금(withdraw), 저금(deposit)
- SavingsAccount(저축예금): 이자율(interest_rate), 이자 더하기(add_interest) **추가**
- CheckingAccount(당좌예금): 수수료(withdraw_charge) 추가, withdraw()를 **재정의**해서 수수료를 더해 출금

```python
class BankAccount:
    def __init__(self, name, number, balance):
        self.balance = balance
        self.name = name
        self.number = number

    def withdraw(self, amount):
        self.balance -= amount
        return self.balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

class SavingsAccount(BankAccount):
    def __init__(self, name, number, balance, interest_rate):
        super().__init__(name, number, balance)
        self.interest_rate = interest_rate

    def set_interest_rate(self, interest_rate):
        self.interest_rate = interest_rate

    def get_interest_rate(self):
        return self.interest_rate

    def add_interest(self):         # 예금에 이자를 더한다.
        self.balance += self.balance * self.interest_rate

class CheckingAccount(BankAccount):
    def __init__(self, name, number, balance):
        super().__init__(name, number, balance)
        self.withdraw_charge = 10000    # 수표 발행 수수료

    def withdraw(self, amount):
        return super().withdraw(amount + self.withdraw_charge)

def test_account():
    a1 = SavingsAccount("홍길동", 123456, 10000, 0.05)
    a1.add_interest()
    print("저축예금의 잔액=", a1.balance)

    a2 = CheckingAccount("김철수", 123457, 2000000)
    a2.withdraw(100000)
    print("당좌예금의 잔액=", a2.balance)

if __name__ == "__main__":
    test_account()
```

# 다형성
- 다형성(polymorphism)은 "많은(poly) + 모양(morph)"이라는 의미로, 하나의 식별자로 다양한 타입(클래스)을 처리하는 것
- 같은 `s.getArea()` 호출이라도 `s`가 어떤 클래스의 객체인지에 따라 다른 코드가 실행된다. (동적 바인딩)
- 부모 클래스에서는 `raise NotImplementedError`로 "자식이 반드시 구현해야 하는 메소드"임을 표시할 수 있다.
- 내장 함수 `len()`도 리스트, 문자열, 딕셔너리에 모두 쓸 수 있는 다형성의 예

```python
class Shape:
    def __init__(self, name):
        self.name = name

    def getArea(self):
        raise NotImplementedError("이것은 추상메소드입니다.")

class Circle(Shape):
    def __init__(self, name, radius):
        super().__init__(name)
        self.radius = radius

    def getArea(self):
        return 3.141592 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, name, width, height):
        super().__init__(name)
        self.width = width
        self.height = height

    def getArea(self):
        return self.width * self.height

def test_polymorphism():
    shapeList = [Circle("c1", 10), Rectangle("r1", 10, 10)]
    for s in shapeList:
        print(s.name, s.getArea())

    try:
        Shape("s1").getArea()
    except NotImplementedError as e:
        print("오류:", e)

    print("리스트의 길이=", len([1, 2, 3]))
    print("문자열의 길이=", len("This is a sentense"))
    print("딕셔너리의 길이=", len({'aaa': 1, 'bbb': 2}))

if __name__ == "__main__":
    test_polymorphism()
```

# 클래스 관계: is-a 와 has-a
- is-a 관계: **상속**
  - 승용차는 차량의 일종이다(Car is a Vehicle).
  - 강아지는 동물의 일종이다(Dog is an animal).
- has-a 관계: **구성** (다른 클래스의 객체를 속성으로 가진다)
  - 도서관은 책을 가지고 있다(Library has a book).
  - 거실은 소파를 가지고 있다(Living room has a sofa).

```python
class Animal(object):
    pass

class Dog(Animal):              # is-a: Dog는 Animal이다.
    def __init__(self, name):
        self.name = name

class Person(object):
    def __init__(self, name):
        self.name = name
        self.pet = None         # has-a: Person은 pet을 가진다.

dog1 = Dog("dog1")
person1 = Person("홍길동")
person1.pet = dog1
print(f"{person1.name}의 애완동물: {person1.pet.name}")
```

# Lab: Card와 Deck
- has-a 관계의 예: Deck은 52개의 Card 객체를 가지고 있다.
- 각 클래스의 `__str__()`을 구현해서 덱 안의 카드를 출력

```python
class Card:
    suitNames = ['클럽', '다이아몬드', '하트', '스페이드']
    rankNames = [None, '에이스', '2', '3', '4', '5', '6', '7',
                 '8', '9', '10', '잭', '퀸', '킹']

    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank

    def __str__(self):
        return Card.suitNames[self.suit] + " " + Card.rankNames[self.rank]

class Deck:
    def __init__(self):
        self.cards = []
        for suit in range(4):
            for rank in range(1, 14):
                card = Card(suit, rank)
                self.cards.append(card)

    def __str__(self):
        lst = [str(card) for card in self.cards]
        return str(lst)

deck = Deck()                   # 덱 객체를 생성한다.
print(len(deck.cards))
print(deck)                     # 덱 객체를 출력한다. __str__()이 호출된다.
```

# Lab: 학생과 강사
- Person(이름, 주민번호)을 상속받아서 Student와 Teacher를 정의
- Student: 수강과목, 평점 / Teacher: 강의과목, 월급

```python
class Person:
    def __init__(self, name, number):
        self.name = name
        self.number = number

    def __str__(self):
        return f"이름={self.name}\n주민번호={self.number}"

class Student(Person):
    UNDERGRADUATE = 0
    POSTGRADUATE = 1

    def __init__(self, name, number, studentType):
        super().__init__(name, number)
        self.studentType = studentType
        self.gpa = 0
        self.classes = []

    def enrollCourse(self, course):
        self.classes.append(course)

    def __str__(self):
        return f"{super().__str__()}\n수강과목={self.classes}\n평점={self.gpa}"

class Teacher(Person):
    def __init__(self, name, number):
        super().__init__(name, number)
        self.courses = []
        self.salary = 3000000

    def assignTeaching(self, course):
        self.courses.append(course)

    def __str__(self):
        return f"{super().__str__()}\n강의과목={self.courses}\n월급={self.salary}"

def test_student_teacher():
    hong = Student("홍길동", "12345678", Student.UNDERGRADUATE)
    hong.enrollCourse("자료구조")
    print(hong)

    kim = Teacher("김철수", "123456790")
    kim.assignTeaching("Python")
    print(kim)

if __name__ == "__main__":
    test_student_teacher()
```

# Lab: Vehicle과 Car, Truck
- 일반적인 운송수단 Vehicle을 상속받아서 Car와 Truck을 작성
- `drive()`, `stop()`은 Vehicle에서는 구현하지 않고 자식 클래스에서 재정의

```python
class Vehicle:
    def __init__(self, name):
        self.name = name

    def drive(self):
        raise NotImplementedError("이것은 추상메소드입니다.")

    def stop(self):
        raise NotImplementedError("이것은 추상메소드입니다.")

class Car(Vehicle):
    def drive(self):
        return '승용차를 운전합니다.'

    def stop(self):
        return '승용차를 정지합니다.'

class Truck(Vehicle):
    def drive(self):
        return '트럭을 운전합니다.'

    def stop(self):
        return '트럭을 정지합니다.'

def test_vehicle():
    cars = [Truck('truck1'), Truck('truck2'), Car('car1')]
    for car in cars:
        print(car.name + ': ' + car.drive())

if __name__ == "__main__":
    test_vehicle()
```

# 추상 클래스와 추상 메소드
- `abc` 모듈의 `ABC`(Abstract Base Class)를 상속받고, 메소드 위에 `@abstractmethod` 데코레이터를 붙여서 정의
- 추상 메소드는 반드시 자식 클래스에서 구현해야 한다.
- 추상 메소드를 하나라도 가진 클래스는 추상 클래스이고, 직접 객체를 만들 수 없다.
- `raise NotImplementedError` 방식은 메소드를 **호출할 때** 오류가 나지만, `ABC` 방식은 **객체를 만들 때** 오류가 난다.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        return "멍멍"

try:
    a = Animal()                # 에러: 추상 클래스는 인스턴스화 불가
except TypeError as e:
    print("오류:", e)

d = Dog()
print(d.sound())
```

# 정적 멤버 (클래스 변수, 정적 메소드)
- 클래스 변수: 클래스 안에 `self` 없이 선언한 변수. 모든 객체가 공유하고 `클래스.변수` 형태로 접근
- 정적 메소드: `@staticmethod` 데코레이터를 붙이고 `self` 매개변수가 없다. 객체 없이 `클래스.메소드()`로 호출

```python
class Counter:
    count = 0                       # 클래스 변수(정적 멤버)

    def __init__(self):
        self.__object_counter = 1   # private 인스턴스 변수

    @staticmethod
    def increment_class():
        Counter.count += 1

    def increment_object(self):
        self.__object_counter += 1

    def get_object_counter(self):
        return self.__object_counter

if __name__ == "__main__":
    Counter.increment_class()       # 객체 없이 호출
    print(Counter.count)

    c = Counter()                   # 인스턴스 생성
    c.increment_object()
    print(c.get_object_counter())
    try:
        print(c.__object_counter)   # private 접근 불가
    except AttributeError as e:
        print("오류:", e)
```

# 연습문제 (추상클래스 다루기)
- 동물(Animal)을 울음(sound)을 추상 메소드로 하는 추상 클래스로 정의
- 개(Dog)와 고양이(Cat)를 구체적인 클래스로 정의하고 각 동물의 울음소리를 출력
- 각 동물 객체는 자신의 이름을 속성으로 가지며 생성자에서 이름을 초기화
- Animal 클래스는 현재 생성된 동물 객체의 수를 클래스 변수로 가지고, 정적 메소드로 그 수를 반환
- 객체가 생성(`__init__`)될 때 수를 증가하고 삭제(`__del__`)될 때 수를 감소

테스트
- 개나 고양이 객체를 Animal 변수로 받아 동적 바인딩이 되는지 확인
- 모든 객체들이 울게 하는 함수를 작성하여 테스트

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    __animal_count = 0              # 현재 동물 객체의 수 (클래스 변수)

    def __init__(self, name: str):
        self.name = name
        Animal.__animal_count += 1

    def __del__(self):
        Animal.__animal_count -= 1

    @staticmethod
    def get_animal_count() -> int:
        return Animal.__animal_count

    @abstractmethod
    def sound(self) -> str:
        pass

    def __repr__(self):
        return f"{type(self).__name__}({self.name})"

class Dog(Animal):
    def sound(self) -> str:
        return "멍멍"

class Cat(Animal):
    def sound(self) -> str:
        return "야옹"

def animal_sound(animal: Animal):
    print(f"{animal.name}: {animal.sound()}")

def animal_sounds(animals: list[Animal]):
    for animal in animals:
        animal_sound(animal)

def test_animal():
    dog: Dog = Dog("바둑이")
    print(dog.sound())
    cat = Cat("야옹이")
    print(cat.sound())
    any1: Animal = Dog("바둑이2")       # Animal 변수로 받아도 Dog의 sound()가 실행된다.
    print(any1.sound())
    any2: Animal = Cat("야옹이2")
    print(any2.sound())
    alist: list[Animal] = [Dog("바둑이a"), Cat("나비b"), Dog("멍멍이c"), Cat("야옹이d")]
    print(alist)

    animal_sound(alist[2])
    animal_sounds(alist)
    print(f"총 동물의 수: {Animal.get_animal_count()}")
    del dog
    del cat
    print(f"총 동물의 수: {Animal.get_animal_count()}")

if __name__ == "__main__":
    test_animal()
```

# Dunder methods
- Dunder = Double Underlines 의 약자. `__XXX__` 형태의 메소드
- 특정 상황에서 자동으로 호출되며, 재정의해서 원하는 코드를 실행하게 할 수 있다.

| 메소드 | 자동으로 호출되는 때 |
|---|---|
| `__init__(self, ...)` | 객체가 만들어질 때: `a = A()` |
| `__str__(self)` | 객체를 문자열로 바꿀 때: `print(a)` |
| `__eq__(self, other)` | 같은지 비교할 때: `a == b` |
| `__gt__(self, other)` | 큰지 비교할 때: `a > b` |
| `__del__(self)` | 객체가 삭제될 때: `del a` |

`dir(object)`로 object 클래스가 가진 메소드 목록을 볼 수 있다.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Student(name={self.name}, age={self.age})"

    def __eq__(self, other):
        if isinstance(other, Student):
            return self.name == other.name and self.age == other.age
        return False

    def __gt__(self, other):
        return self.age > other.age

    def __del__(self):
        print(f"Student {self.name} is being deleted")

def test_dunder():
    student1 = Student("Alice", 20)
    student2 = Student("Bob", 22)
    student3 = Student("Alice", 20)

    print(student1)                 # __str__
    print(student2)                 # __str__
    print(student1 == student2)     # __eq__
    print(student1 == student3)     # __eq__
    print(student2 > student1)      # __gt__

    del student1                    # __del__
    print("테스트 끝")              # 함수가 끝나면 남은 지역 변수도 삭제된다.

if __name__ == "__main__":
    test_dunder()
```

# 데코레이터
- 함수나 클래스의 동작을 수정하거나 확장할 때 사용하는 문법
- 함수 정의 위에 `@데코레이터이름` 형태로 사용
- 데코레이터는 함수를 받아서, 그 함수를 감싼 새 함수(wrapper)를 반환하는 함수
- `@staticmethod`, `@abstractmethod`도 데코레이터

이미 작성된 함수를 확장하는 2가지 방법
1. 함수 재할당: `원본함수명 = decorator(원본함수명)` → 기존에 호출하던 코드를 바꾸지 않아도 됨 (`@` 문법과 같은 효과)
2. 다른 함수명: `신규함수명 = decorator(원본함수명)` → 원본은 그대로 두고 새 이름으로 사용

```python
def simple_decorator(func):
    def wrapper(*args, **kwargs):
        print("함수 실행 전")       # 추가적인 코드
        result = func(*args, **kwargs)  # 목적 함수 실행
        print("함수 실행 후")       # 추가적인 코드
        return result
    return wrapper

@simple_decorator
def say_hello():
    print("Hello, World!")

def say_bye(name):
    print(f"Bye, {name}!")

decorated_say_bye = simple_decorator(say_bye)   # 2. 다른 함수명

if __name__ == "__main__":
    print("=== @ 문법 ===")
    say_hello()
    print("=== 원본 함수 ===")
    say_bye("Python")
    print("=== 다른 함수명 ===")
    decorated_say_bye("Python")
```

- 상속은 다른 클래스를 재사용하는 방법이다. 객체 간에 is-a 관계가 성립하면 상속을 이용한다.
- 상속을 사용하면 중복된 코드를 줄일 수 있다. 공통적인 코드는 부모 클래스 한 곳으로 모은다.
- 자식 클래스의 생성자에서는 `super().__init__()`으로 부모 클래스의 생성자를 호출해야 한다.
- 부모 클래스의 메소드를 자식 클래스가 재정의하는 것을 메소드 오버라이딩이라고 한다.
- 같은 이름의 메소드가 객체의 클래스에 따라 다르게 동작하는 것이 다형성이다.
- has-a 관계는 상속이 아니라 구성(객체를 속성으로 가짐)으로 표현한다.
