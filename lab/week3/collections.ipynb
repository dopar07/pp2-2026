## tuple, set, dictionary 실습

- tuple
변경이 불가능한 리스트

```python
tuple1 = (1, 2, 3)
```

- set
중복이 불가능하고 순서가 없는 집합

```python
set1 = {1, 2, 3}
```

- dictionary
키와 벨류로 된 항목들의 집합

```python
dict1 = {'a': 1, 'b': 2, 'c': 3}
```

byte, bytes, bytearray 공부
- bytes
불변, 바이트 버전에 튜플
- bytearray
가변, 바이트 버전에 리스트

```python
# 1. 문자열 리터럴 앞에 b 붙이기 (ASCII 문자만 가능)
b1 = b'hello'

# 2. 지정한 길이만큼 0으로 채워진 객체 생성 (5바이트)
b2 = bytes(5)  # b'\x00\x00\x00\x00\x00'

# 3. 0~255 사이의 정수 리스트를 바이트로 변환
b3 = bytes([72, 101, 108, 108, 111])  # b'Hello'

# 4. 일반 문자열(str)을 인코딩하여 생성
b4 = '안녕'.encode('utf-8')

ba1 = bytearray(b'hello')
ba2 = bytearray(5)
ba3 = bytearray([72, 101, 108, 108, 111])
```

all(), any() 공부
- all()
모든 조건이 참이여야 참을 출력(and)
- any()
하나의 조건만 참이면 참을 출력(or)
